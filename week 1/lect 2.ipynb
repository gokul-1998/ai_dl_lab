# 14)  get the l1 norm and l2 norm

- what is l1 norm?  
    - it is the sum of absolute values of the vector elements. It is also known as Manhattan norm or Taxicab norm.
    - the formula is: ||x||1 = |x1| + |x2| + ... + |xn|
    - eg: for vector [3, -4, 5], l1 norm = |3| + |-4| + |5| = 3 + 4 + 5 = 12
- what is l2 norm?
    - it is the square root of the sum of the squares of the vector elements. It is also known as Euclidean norm.
    - the formula is: ||x||2 = sqrt(x1^2 + x2^2 + ... + xn^2)
    - eg: for vector [3, -4, 5], l2 norm = sqrt(3^2 + (-4)^2 + 5^2) = sqrt(9 + 16 + 25) = sqrt(50) = 5.0
