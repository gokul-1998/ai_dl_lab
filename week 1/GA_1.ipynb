```python
import pandas as pd

# Load dataset
df = pd.read_csv("V5.csv")

# --- Q12 ---
even_rows = df.iloc[::2, :]
q12_value = even_rows.iloc[332, 3]

# --- Q13 ---
odd_rows = df.iloc[1::2, :]
q13_value = odd_rows.iloc[332, 3]

# --- Q14 ---
odd_cols = df.iloc[:, 1::2]
q14_value = odd_cols.iloc[100, 5]

# --- Q15 ---
even_cols_rows = df.iloc[::2, ::2]
q15_value = even_cols_rows.iloc[255, 3]

# --- Q16 ---
most_recent_years = sorted(df["Year"].unique())[-6:]
q16_value = df[df["Year"].isin(most_recent_years)].shape[0]

# --- Q17 ---
q17_value = df[(df["num_rooms"] == 3) & (df["num_bathrooms"] == 3)].shape[0]

# --- Q18 ---
q18_value = df[(df["num_rooms"] == 3) | (df["num_bathrooms"] == 3)].shape[0]

# --- Q19 ---
q19_value = df[(df["Year"] == 2022) & 
               (df["Locality"] == "Greenwich") & 
               (df["num_rooms"] == 3) & 
               (df["Face"].isin(["North", "East"]))].shape[0]

# --- Q20 ---
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
q20_value = df[df["Date"].dt.month_name() == "August"].shape[0]

# --- Q21 ---
q21_value = df.groupby("Locality")["Sale Price"].mean().idxmax()

# --- Print all answers ---
print("Q12:", q12_value)
print("Q13:", q13_value)
print("Q14:", q14_value)
print("Q15:", q15_value)
print("Q16:", q16_value)
print("Q17:", q17_value)
print("Q18:", q18_value)
print("Q19:", q19_value)
print("Q20:", q20_value)
print("Q21:", q21_value)
```
